```python
import tkinter as tk
import customtkinter as ctk
from deep_translator import GoogleTranslator
import pyperclip
import threading
from gtts import gTTS
from playsound import playsound
import os

ctk.set_appearance_mode("dark")
ctk.set_default_color_theme("blue")


class TranslationApp(ctk.CTk):
    def __init__(self):
        super().__init__()

        self.title("🌍 AI Translator Pro")
        self.geometry("800x600")

        self.languages = {
            "English": "en", "Spanish": "es", "French": "fr",
            "German": "de", "Hindi": "hi", "Arabic": "ar",
            "Chinese": "zh-CN", "Japanese": "ja"
        }

        self.setup_ui()

    def setup_ui(self):
        title = ctk.CTkLabel(self, text="🌍 AI Language Translator",
                             font=ctk.CTkFont(size=28, weight="bold"))
        title.pack(pady=15)

        container = ctk.CTkFrame(self, corner_radius=15)
        container.pack(fill="both", expand=True, padx=20, pady=10)

        # LEFT
        left = ctk.CTkFrame(container)
        left.pack(side="left", expand=True, fill="both", padx=10, pady=10)

        self.src_lang = ctk.CTkComboBox(left, values=list(self.languages.keys()))
        self.src_lang.set("English")
        self.src_lang.pack(fill="x", pady=5)

        self.input_box = ctk.CTkTextbox(left)
        self.input_box.pack(expand=True, fill="both")

        # CENTER
        center = ctk.CTkFrame(container, fg_color="transparent")
        center.pack(side="left")

        ctk.CTkButton(center, text="⇄", command=self.swap_languages).pack(pady=200)

        # RIGHT
        right = ctk.CTkFrame(container)
        right.pack(side="right", expand=True, fill="both", padx=10, pady=10)

        self.tgt_lang = ctk.CTkComboBox(right, values=list(self.languages.keys()))
        self.tgt_lang.set("Hindi")
        self.tgt_lang.pack(fill="x", pady=5)

        self.output_box = ctk.CTkTextbox(right)
        self.output_box.pack(expand=True, fill="both")
        self.output_box.configure(state="disabled")

        # BUTTONS
        btn_frame = ctk.CTkFrame(self, fg_color="transparent")
        btn_frame.pack(fill="x", padx=20, pady=10)

        ctk.CTkButton(btn_frame, text="🚀 Translate",
                      command=self.thread_translate).pack(side="left", expand=True, padx=5)

        ctk.CTkButton(btn_frame, text="📋 Copy",
                      command=self.copy_text).pack(side="left", expand=True, padx=5)

        # 🔊 NEW SPEAK BUTTON
        ctk.CTkButton(btn_frame, text="🔊 Speak",
                      command=self.thread_speak).pack(side="left", expand=True, padx=5)

        self.status = ctk.CTkLabel(self, text="Ready", anchor="w")
        self.status.pack(fill="x", padx=10, pady=5)

    # ------------------------
    # THREAD HANDLERS
    # ------------------------

    def thread_translate(self):
        threading.Thread(target=self.translate_text, daemon=True).start()

    def thread_speak(self):
        threading.Thread(target=self.speak_text, daemon=True).start()

    # ------------------------
    # CORE FUNCTIONS
    # ------------------------

    def translate_text(self):
        text = self.input_box.get("1.0", tk.END).strip()

        if not text:
            self.update_status("⚠️ Enter text")
            return

        self.update_status("⏳ Translating...")

        try:
            result = GoogleTranslator(
                source=self.languages[self.src_lang.get()],
                target=self.languages[self.tgt_lang.get()]
            ).translate(text)

            self.output_box.configure(state="normal")
            self.output_box.delete("1.0", tk.END)
            self.output_box.insert("1.0", result)
            self.output_box.configure(state="disabled")

            self.update_status("✅ Done")

        except Exception as e:
            print("Translation Error:", e)
            self.update_status("❌ Error")

    def speak_text(self):
        text = self.output_box.get("1.0", tk.END).strip()

        if not text:
            self.update_status("⚠️ No text")
            return

        try:
            self.update_status("🔊 Speaking...")

            lang_code = self.languages[self.tgt_lang.get()]

            tts = gTTS(text=text, lang=lang_code)
            filename = "voice.mp3"
            tts.save(filename)

            playsound(filename)

            os.remove(filename)

            self.update_status("✅ Done")

        except Exception as e:
            print("Speech Error:", e)
            self.update_status("❌ Speech error")

    def copy_text(self):
        text = self.output_box.get("1.0", tk.END).strip()
        if text:
            pyperclip.copy(text)
            self.update_status("📋 Copied")

    def swap_languages(self):
        src = self.src_lang.get()
        tgt = self.tgt_lang.get()
        self.src_lang.set(tgt)
        self.tgt_lang.set(src)

    def update_status(self, msg):
        self.status.configure(text=msg)


if __name__ == "__main__":
    app = TranslationApp()
    app.mainloop()
```
